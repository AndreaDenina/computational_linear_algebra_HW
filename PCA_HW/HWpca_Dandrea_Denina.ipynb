#### Describe and motivate the encoding operations you will apply (max 150 words):
Our subset contains three categorical features: `Smoking`, `Punctuality` and `Internet usage`. Since the assignment states that all categorical values are ordered, we encoded them with an `OrdinalEncoder`, passing the order explicitly. We did not use one-hot encoding because it would lose this order and turn each question into 3 or 4 columns: after scaling, that question would weigh 3 or 4 times more than the others, and its columns, always linked to each other, would show PCA a false pattern.

Ordinal encoding instead keeps one column per question, with values from 1 to 3 (or 4), like the 1 to 5 ratings. This assumes equal spacing between consecutive levels, as the ratings already do. For `Smoking`, we ordered the levels by degree of smoking habit: tried smoking implies no habit, while a former smoker had a regular one, so it ranks just below current smokers.

#### Describe the preprocessing operation you will apply and comment the effects it may have on the data (max 150 words):
We applied the `StandardScaler`, which transforms each feature $j$ as
$$z_{ij} = \frac{x_{ij}-\mu_j}{\sigma_j}, \qquad \sigma_j^2=\frac{1}{N}\sum_{i=1}^{N}(x_{ij}-\mu_j)^2,$$
where $N=506$ is the number of people. Each feature gets mean 0 and unit variance, so the covariance matrix becomes the correlation matrix $R$:
$$\text{Cov}(z_j,z_k)=\frac{\text{Cov}(x_j,x_k)}{\sigma_j\,\sigma_k}=\rho_{jk}.$$

**Pros.** PCA ignores the spread of each feature and looks only at correlations. The variance along a unit direction $w$ is
$$w^T R\, w = 1+\sum_{j\neq k}w_j\,w_k\,\rho_{jk},$$
so any variance above 1 comes only from groups of correlated questions, which is what we need to identify profiles. High-variance features (e.g., `Pets`) cannot dominate the first PCs, and the different ranges of the encoded features (1 to 3, 1 to 4) no longer matter.

**Cons.** Low-variance features are amplified together with their noise, and rare answers (e.g., the few 1s in `Music`) get large values. The information carried by the original variances (which questions split people the most) is lost, and, with the variance spread evenly, more PCs are needed to reach the same share.

```python
# [write the code here]
# categorical feats: Smoking, Punctuality, Internet usage


cat_features = ['Smoking', 'Punctuality', 'Internet usage']

for cat in cat_features:
    print(f"\n{cat} values:")
    for e in responses_ft[cat].unique():
        print(e)

cat_orders = [
    ['never smoked', 'tried smoking', 'former smoker', 'current smoker'],
    ['early', 'on time', 'late'],
    ['less than an hour a day', 'few hours a day', 'most of the day'],
]

ordinal_enc = OrdinalEncoder(categories=cat_orders)

responses_ft_enc = responses_ft.copy()
# +1 to conform to the other feats, int for the same reason
responses_ft_enc[cat_features] = (ordinal_enc.fit_transform(
    responses_ft[cat_features]) + 1).astype(int)

display(responses_ft_enc)

std_sc = StandardScaler()
responses_ft_pp = pd.DataFrame(std_sc.fit_transform(responses_ft_enc),
                               columns=responses_ft_enc.columns,
                               index=responses_ft_enc.index)

display(responses_ft_pp)
display(responses_ft_pp.describe().loc[['mean', 'std']])   # mean about 0, std about 1.001
print("Total variance (encoded):", round(responses_ft_enc.var().sum(), 2))   # 133.2
print("Total variance (scaled): ", round(responses_ft_pp.var().sum(), 2))    # 93.18
print("Average variance (scaled)", responses_ft_pp.var().mean())

print("Average mean (encoded):", responses_ft_enc.mean().mean())   
print("Average mean (scaled): ", responses_ft_pp.mean().mean())    
```

#### Comment the results obtained after the preprocessing operation (max 100 words):
After scaling, all 93 features have mean ≈ 0 and variance ≈ 1.002. The scaler divides by $N$, while pandas computes the variance as
$$s_j^2=\frac{1}{N-1}\sum_{i=1}^{N}(x_{ij}-\bar{x}_j)^2,$$
so each column shows $N/(N-1)\approx1.002$. The total variance drops from 133.2 in `responses_ft_enc` to 93.18, now equally shared among the features instead of ranging from 0.39 (`Music`) to 2.43 (`Pets`). `Pets` goes from [1, 5] to [-1.49, 1.08], while the near-unanimous `Music` (82% answered 5) goes to [-6.03, 0.41]: its rare answers became extreme values, as expected.
